```python
import pandas as pd
import numpy as np

# Create sample sales data
np.random.seed(42)
n = 500

dates = pd.date_range("2025-01-01", "2025-12-31", freq="D")

df = pd.DataFrame({
    "Order_Date": np.random.choice(dates, n),
    "Region": np.random.choice(
        ["South", "North", "East", "West"], n
    ),
    "Category": np.random.choice(
        ["Technology", "Furniture", "Office Supplies"], n
    ),
    "Product": np.random.choice(
        ["Laptop", "Chair", "Phone", "Desk", "Printer"], n
    ),
    "Quantity": np.random.randint(1, 10, n),
    "Unit_Price": np.random.randint(100, 2000, n),
    "Discount": np.random.choice(
        [0, 0.05, 0.10, 0.15], n
    )
})

df["Sales"] = (
    df["Quantity"] * df["Unit_Price"] *
    (1 - df["Discount"])
).round(2)

df["Profit"] = (
    df["Sales"] * np.random.uniform(0.05, 0.25, n)
).round(2)

df.head()
```

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = len(df)
avg_order_value = total_sales / total_orders

print("SALES DASHBOARD SUMMARY")
print("-" * 30)
print(f"Total Sales: ₹{total_sales:,.2f}")
print(f"Total Profit: ₹{total_profit:,.2f}")
print(f"Total Orders: {total_orders:,}")
print(f"Average Order Value: ₹{avg_order_value:,.2f}")
```

```python
import plotly.express as px
from IPython.display import display, Markdown

# Monthly sales trend
df["Month"] = df["Order_Date"].dt.to_period("M").astype(str)
monthly = df.groupby("Month", as_index=False)["Sales"].sum()

fig1 = px.line(
    monthly,
    x="Month",
    y="Sales",
    markers=True,
    title="Monthly Sales Trend"
)
fig1.show()

# Sales by region
region_sales = df.groupby("Region", as_index=False)["Sales"].sum()

fig2 = px.bar(
    region_sales,
    x="Region",
    y="Sales",
    color="Region",
    title="Sales by Region"
)
fig2.show()

# Sales by category
category_sales = df.groupby("Category", as_index=False)["Sales"].sum()

fig3 = px.pie(
    category_sales,
    names="Category",
    values="Sales",
    title="Sales Share by Category"
)
fig3.show()

# Top 5 products
product_sales = (
    df.groupby("Product", as_index=False)["Sales"]
    .sum()
    .sort_values("Sales", ascending=False)
    .head(5)
)

fig4 = px.bar(
    product_sales,
    x="Product",
    y="Sales",
    title="Top Products by Sales"
)
fig4.show()
```

```python
# Find the best-performing region and product
best_region = region_sales.loc[
    region_sales["Sales"].idxmax()
]

best_product = product_sales.iloc[0]

best_month = monthly.loc[
    monthly["Sales"].idxmax()
]

# Compare the first and second halves of the year
first_half = df[df["Order_Date"].dt.month <= 6]["Sales"].sum()
second_half = df[df["Order_Date"].dt.month > 6]["Sales"].sum()

display(Markdown("## 🤖 Automated Business Insights"))

print(
    f"1. Top region: {best_region['Region']} "
    f"with sales of ₹{best_region['Sales']:,.2f}."
)

print(
    f"2. Best product among the top five: "
    f"{best_product['Product']} "
    f"with sales of ₹{best_product['Sales']:,.2f}."
)

print(
    f"3. Highest-sales month: {best_month['Month']} "
    f"with ₹{best_month['Sales']:,.2f}."
)

if second_half > first_half:
    print("4. Total sales were higher in July–December.")
else:
    print("4. Total sales were higher in January–June.")

print(
    "5. Investigate profit margins, discount levels and "
    "regional demand before making business decisions."
)
```

```python
# Save the generated dataset
df.to_csv("sales_data.csv", index=False)

# Create a simple project summary
summary = pd.DataFrame({
    "Metric": [
        "Total Sales",
        "Total Profit",
        "Total Orders",
        "Average Order Value"
    ],
    "Value": [
        round(total_sales, 2),
        round(total_profit, 2),
        total_orders,
        round(avg_order_value, 2)
    ]
})

summary.to_csv("sales_summary.csv", index=False)

print("Files created successfully!")
print("1. sales_data.csv")
print("2. sales_summary.csv")
```
